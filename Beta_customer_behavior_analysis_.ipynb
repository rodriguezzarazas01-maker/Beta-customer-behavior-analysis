Como primer paso del análisis, se depuró la información básica de los clientes mediante la normalización de nombres y la estandarización de variables demográficas. Con el objetivo de mejorar la eficiencia del proceso de preparación de datos, se desarrolló una función encargada de aplicar de manera uniforme las reglas de limpieza sobre cada registro de cliente. Este procedimiento permitió estandarizar nombres, edades y categorías de compra, garantizando mayor consistencia en la información y reduciendo el riesgo de errores durante análisis posteriores de segmentación y comportamiento del consumidor.

```python

def clean_user(user):
    # Limpia y divide el nombre
    name = user[1].strip().replace('_', ' ').split()

    # Edad como entero
    age = int(user[2])

    # Categorías en minúsculas, tomadas directamente de user[3]
    categories = list(map(str.lower, user[3]))

    return [user[0], name, age, categories, user[4]]


# Prueba
test_user = ['32415', ' mike_reed ', 32.0, ['ELECTRONICS', 'SPORT', 'BOOKS'], [894, 213, 173]]
print(clean_user(test_user))

```
